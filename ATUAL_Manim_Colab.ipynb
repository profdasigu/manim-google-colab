```python
# ============================================================
# ETAPA 5 — CRIAÇÃO DA ANIMAÇÃO
# ============================================================
# Nesta etapa, escrevemos o código da animação que será
# produzida pelo Manim.
# ============================================================


class MinhaCena(Scene):
    # Cria uma cena chamada "MinhaCena".
    #
    # Scene é uma classe do Manim utilizada como base
    # para a construção das animações.


    def construct(self):
        # O método construct() contém os elementos e as
        # animações que serão apresentados na cena.


        titulo = Text("Minha primeira animação!")
        # Cria um objeto de texto.


        self.play(Write(titulo))
        # Anima a escrita do texto na tela.


        self.wait(2)
        # Mantém a cena parada durante 2 segundos.
```

```python
# ============================================================
# ETAPA 6 — RENDERIZAÇÃO DA ANIMAÇÃO
# ============================================================
# Nesta etapa, solicitamos ao Manim que transforme a cena
# criada anteriormente em um vídeo.
# ============================================================


%%manim -qm MinhaCena
# %%manim → executa o Manim diretamente no Google Colab.
#
# -qm → define a renderização em qualidade média
#       (medium quality).
#
# MinhaCena → corresponde ao nome da classe criada
#             na ETAPA 5.
```
